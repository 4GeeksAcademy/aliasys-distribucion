# Problemas de distribución de probabilidad

```python
# Importar las librerías necesarias
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from scipy.stats import norm, uniform
```

## Distribución binomial

En teoría de la probabilidad y estadística, la distribución binomial es la distribución de probabilidad discreta que da sólo dos resultados posibles en un experimento; por ejemplo, al lanzar una moneda, puede salir cara o cruz.

```python
# Definir la función de visualización de la distribución binomial
# Definir todas las funciones de probabilidad relacionadas con esta distribución.

def pmf_binomial(x, size, prob = 0.5):
    """
    Calcula la estimación puntual de la distribución binomial.
    """
    result = binom.pmf(k = x, n = size, p = prob, loc = 0)
    return result


def cdf_binomial(q, size, prob = 0.5):
    """
    Calcula el acumulado de la distribución binomial.
    """
    result = binom.cdf(k = q, n = size, p = prob, loc = 0)
    return result


def ppf_binomial(p, size, prob = 0.5):
    """
    Calcula la función cuantil a partir de la distribución binomial.
    """
    result = binom.ppf(q = p, n = size, p = prob, loc = 0)

    return result


def rvs_binomial(n, size, prob = 0.5):
    """
    Genera variables aleatorias a partir de la distribución binomial.
    """
    result = binom.rvs(n = size, p = prob, size = n)
    return result
```

```python
# Distribución binomial(10, 0.2), dónde n = 10 es el número de ensayos y p = 0.2 es la probabilidad de éxito en un solo ensayo

np.random.seed(42)

print(f"Probability that a Binomial(10, 0.2) takes the value 2: {pmf_binomial(2, size = 10, prob = 0.2)}")
print(f"Probability that a Binomial(10, 0.2) will take a value less than 2: {cdf_binomial(2, size = 10, prob = 0.2)}")
print(f"Which value of a Binomial(10, 0.2) has a cumulative probability of 0.9? {ppf_binomial(0.9, size = 10, prob = 0.2)}")
print(f"Generate 2000 random values from a Binomial(10, 0.2) distribution: {rvs_binomial(2000, size = 10, prob = 0.2)}")
```

```python
# Visualización de la distribución binomial(10, 0.2)
np.random.seed(42)

size = 10
prob = 0.2
data = rvs_binomial(2000, size = 10, prob = 0.2)

pmf = pmf_binomial(data, size = size, prob = prob)
cdf = cdf_binomial(data, size = size, prob = prob)
ppf = ppf_binomial(data, size = size, prob = prob)

fig, axis = plt.subplots(3, 1, figsize = (5, 7))

axis[0].stem(data, pmf, "r-")
axis[1].stem(data, cdf, "b-")
axis[2].stem(data, ppf, "g-")

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

# Distribucion normal

```python
# Empezamos con las distribuciones normales
def pdf_normal(x, mean = 0, sd = 1):
    """
    Calcula la estimación puntual de la distribución normal.
    """
    result = norm.pdf(x, loc = mean, scale = sd)
    return result


def cdf_normal(q, mean = 0, sd = 1):
    """
    Calcula el acumulado de la distribución normal.
    """
    result = norm.cdf(q, loc = mean, scale = sd)
    return result


def ppf_normal(p, mean = 0, sd = 1):
    """
    Calcula la función cuantil a partir de la distribución normal.
    """
    result = norm.ppf(p, loc = mean, scale = sd)
    return result


def rvs_normal(n, mean = 0, sd = 1):
    """
    Genera variables aleatorias a partir de la distribución normal.
    """
    result = norm.rvs(loc = mean, scale = sd, size = n)
    return result
```

```python
# Distribución normal(0, 1), dónde 0 es la media y 1 es la desviación estándar
np.random.seed(42)
print(f"Probability that a Normal(0, 1) takes the value 2: {pdf_normal(2, mean = 0, sd = 1)}")
print(f"Probability that a Normal(0, 1) will take a value less than 2: {cdf_normal(2, mean = 0, sd = 1)}")
print(f"Which value of a Normal(0, 1) has a cumulative probability of 0.9? {ppf_normal(0.9, mean = 0, sd = 1)}")
print(f"Generate 2000 random values from a Normal(0, 1) distribution: {rvs_normal(2000, mean = 0, sd = 1)}")
```

# Distribucion uniforme

```python
# Calculamos la distribucion uniforme
def pmf_unif(x, min = 0, max = 1):
    """
    Calcula la estimación puntual de la distribución uniforme.
    """
    if min <= x <= max:
        return 1 / (max - min)
    else:
        return 0
    

def cdf_unif(q, min = 0, max = 1):
    """
    Calcula el acumulado de la distribución uniforme.
    """
    if q < min:
        return 0
    elif q > max:
        return 1
    else:
        return (q - min) / (max - min)
    

def ppf_unif(p, min = 0, max = 1):
    """
    Calcula la función cuantil a partir de la distribución uniforme.
    """
    if p < 0 or p > 1:
        raise ValueError("p must be between 0 and 1")
    return min + p * (max - min)


def rvs_unif(n, min = 0, max = 1):
    """
    Genera variables aleatorias a partir de la distribución uniforme.
    """
    return np.random.uniform(low = min, high = max, size = n)
```

```python
# Distribucuión uniforme(0, 1), dónde 0 es el mínimo y 1 es el máximo
np.random.seed(42)
print(f"Probability that a Uniform(0, 1) takes the value 0.5: {pmf_unif(0.5, min = 0, max = 1)}")
print(f"Probability that a Uniform(0, 1) will take a value less than 0.5: {cdf_unif(0.5, min = 0, max = 1)}")
print(f"Which value of a Uniform(0, 1) has a cumulative probability of 0.9? {ppf_unif(0.9, min = 0, max = 1)}")
print(f"Generate 2000 random values from a Uniform(0, 1) distribution: {rvs_unif(2000, min = 0, max = 1)}")
```

```python
# Visualización de la distribución uniforme(0, 1)
np.random.seed(42)

min_val = 0
max_val = 1
data = rvs_unif(2000, min = min_val, max = max_val)

pmf = [pmf_unif(x, min = min_val, max = max_val) for x in data]
cdf = [cdf_unif(q, min = min_val, max = max_val) for q in data]
ppf = [ppf_unif(p, min = min_val, max = max_val) for p in np.linspace(0, 1, 2000)]

fig, axis = plt.subplots(3, 1, figsize = (5, 7))
axis[0].plot(data, pmf, "r-")
axis[1].plot(data, cdf, "b-")
axis[2].plot(data, ppf, "g-")

plt.tight_layout()
plt.show()
```
